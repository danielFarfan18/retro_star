```python
import subprocess
import tempfile
import os

def RSMILES_prediction(model_path, src_str, output_path, gpu_id=-1, beam_size=10, n_best=10, batch_type="tokens", max_length=1000, seed=0):
    # Escribe la cadena de entrada en un archivo temporal
    with tempfile.NamedTemporaryFile(delete=False, mode='w') as f:
        f.write(src_str + '\n')
        src_path = f.name  # Usa la ruta al archivo temporal como el argumento src

    # Construye el comando
    cmd = f"onmt_translate -model {model_path} -src {src_path} -output {output_path} -gpu {gpu_id} -beam_size {beam_size} -n_best {n_best} -batch_type {batch_type} -max_length {max_length} -seed {seed}"

    # Ejecuta el comando
    subprocess.run(cmd, shell=True)

    # Elimina el archivo de origen
    os.remove(src_path)
model_path = "one_step_model/finetuned_model.product-reactants_step_215000.pt"
src_str = "CCCC[C@@H](C(=O)N1CCC[C@H]1C(=O)O)[C@@H](F)C(=O)OC"
output_path = "exp/outputs.txt"
RSMILES_prediction(model_path, src_str, output_path)
```
